FEATURE ENGINEERING
“Creating new smart columns from existing data to help Machine Learning understand patterns better.”



```python
#UTILIZATION RATIO

#Formula
#transactions / total ration cards

final_df['utilization_ratio'] = final_df['noOfTrans'] / final_df['totalRcs']

#Meaning
#Shows: how actively ration cards are being used.
```

```python
#PORTABILITY RATIO
#Formula
#other shop transactions / total transactions

final_df['portability_ratio'] = (
    final_df['otherShopTransCnt'] / final_df['noOfTrans']
)

#Meaning
#Shows: how many beneficiaries are coming from other shops.
#Business Meaning
#High portability shops:
#transport hubs
#urban centers
#high demand areas
```

```python
#RICE DISTRIBUTION RATIO

final_df['rice_distribution_ratio'] = (
    final_df['riceFsc'] / (final_df['totalUnits'] + 1)
)

#Meaning
#Rice distributed per beneficiary.
```

```python
#TRANSACTION GAP

final_df['transaction_gap'] = (
    final_df['totalRcs'] - final_df['noOfTrans']
)

#Meaning

#Difference between:
#eligible cards
#actual usage

#Interpretation

#arge gap may indicate:
#inactive users
#suspicious activity
```

```python
#TOTAL COMMODITY

final_df['total_commodity'] = (
    final_df['riceFsc'] +
    final_df['riceAfsc'] +
    final_df['wheat'] +
    final_df['sugar']
)

#Meaning

#Total commodity distributed by shop.
```

```python
#VOLATILITY FEATURE

volatility = final_df.groupby('shopNo')['noOfTrans'].std()

final_df = final_df.merge(
    volatility.rename('transaction_volatility'),
    on='shopNo',
    how='left'
)

#Meaning

#Measures: how unstable shop transactions are over time.
```

```python
#NOW YOUR ML WILL WORK
features = ['utilization_ratio', 'portability_ratio', 'transaction_gap']

X = df[features]
```

HANDLE INFINITE / NULL VALUES
Very important before ML.

Some divisions may create:

infinity
NaN

ML cannot handle them.

```python
import numpy as np

features = features.replace([np.inf, -np.inf], np.nan)

features = features.fillna(0)
```

STANDARD SCALING

VERY IMPORTANT.

project specifically asked this.

WHAT IS SCALING?

All columns have different ranges.

Example:

transaction_gap → thousands
utilization_ratio → decimals

Scaling makes all columns balanced.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)

#StandardScaler was applied before clustering to normalize feature ranges for distance-based
#algorithms.”
```

```python
#SCALING (VERY IMPORTANT)

DBSCAN + KMeans need scaled data.
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

ELBOW METHOD

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(1, 10):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)

plt.plot(range(1,10), inertia, marker='o')

plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.show()


#Find the “bend” in graph.

#Usually:4 or 5 clusters.
```

APPLY K-MEANS

```python
kmeans = KMeans(n_clusters=4, random_state=42)

final_df['cluster'] = kmeans.fit_predict(scaled_features)
```

CHECK CLUSTER COUNTS

```python
final_df['cluster'].value_counts()
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42)

df['cluster'] = kmeans.fit_predict(X_scaled)
```

SILHOUETTE SCORE

Measures clustering quality.

```python
#Instead of full dataset, use only sample rows.

sample_features = scaled_features[:10000]
sample_clusters = final_df['cluster'][:10000]

from sklearn.metrics import silhouette_score

score = silhouette_score(sample_features, sample_clusters)

print(score)

#Silhouette Score 
# “The K-Means clustering model achieved a silhouette score of 0.357, 
#indicating acceptable separation between behavioral shop clusters.”
```
